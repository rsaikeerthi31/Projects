# Project 9 Voice Assistant

### Project Description
Builds a bi-directional Voice-to-Text and Text-to-Voice AI Pipeline using `SpeechRecognition` and `pyttsx3`.

```python
import speech_recognition as sr
import pyttsx3

class VoiceAssistant:
    def __init__(self):
        self.recognizer = sr.Recognizer()
        self.tts = pyttsx3.init()
        self.tts.setProperty('rate', 150)

    def speak(self, text: str):
        print(f"Assistant: {text}")
        self.tts.say(text)
        self.tts.runAndWait()

    def listen(self):
        with sr.Microphone() as source:
            self.recognizer.adjust_for_ambient_noise(source, duration=1)
            print("Listening...")
            try:
                audio = self.recognizer.listen(source, timeout=5)
                text = self.recognizer.recognize_google(audio)
                print(f"You: {text}")
                return text
            except Exception as e:
                print("Could not process audio.")
                return None

if __name__ == "__main__":
    assistant = VoiceAssistant()
    assistant.speak("Ready.")
    text = assistant.listen()
    if text: assistant.speak(f"You said {text}")
```

### Project Explanation
Uses a `with sr.Microphone()` context manager to safely access hardware. Adjusts for ambient room noise, streams audio to Google's Speech API, and uses offline `pyttsx3` native OS voices to speak the result.
