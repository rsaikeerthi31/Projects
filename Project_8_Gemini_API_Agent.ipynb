# Project 8 Gemini API Agent

### Project Description
Connects to Google's most powerful AI models using the official `google-genai` Python SDK, securely loading API keys from the environment.

```python
import os
import sys
from google import genai
from google.genai import types

def query_gemini(prompt_text: str):
    api_key = os.environ.get("GEMINI_API_KEY")
    if not api_key:
        print("Set GEMINI_API_KEY environment variable.")
        return
        
    try:
        client = genai.Client()
        response = client.models.generate_content(
            model='gemini-2.5-flash',
            contents=prompt_text,
            config=types.GenerateContentConfig(temperature=0.7)
        )
        print(response.text)
    except Exception as e:
        print(f"Error: {e}")

if __name__ == "__main__":
    query_gemini("Explain a list vs tuple in exactly two sentences.")
```

### Project Explanation
Securely fetches `GEMINI_API_KEY` from the OS. Initializes the `genai.Client()` and uses `gemini-2.5-flash` with a temperature config of 0.7 for creative, natural responses.

### Project Remarks
For real-world apps, use `generate_content_stream` to print words one-by-one, reducing perceived latency.
