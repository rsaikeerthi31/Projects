# Project 5 FastAPI Mock Database

### Project Description
Builds a lightning-fast REST API using FastAPI. Simulates a backend database using the Faker module to generate randomized data points, and utilizes URL Query Parameters to limit results.

```python
from fastapi import FastAPI, Query
from faker import Faker

app = FastAPI(title="Mock Corporate API")
fake = Faker()

users_db = [{"id": i, "name": fake.name(), "email": fake.email()} for i in range(1, 201)]
employees_db = [{"id": i, "title": fake.job(), "company": fake.company()} for i in range(1, 201)]

@app.get("/")
def home():
    return {"message": "Welcome", "endpoints": ["/users", "/employees"]}

@app.get("/users")
def get_users(limit: int = Query(200)):
    return users_db[:limit]

@app.get("/employees")
def get_employees(limit: int = Query(200)):
    return employees_db[:limit]

# Run with: uvicorn main:app --reload
```

### Project Explanation
Generates persistent mock data lists using Faker and List Comprehensions at startup. URL queries (e.g., `?limit=10`) are strictly type-validated by FastAPI and applied using Python list slicing.
