# Project 1 Sudoku Solver

### Project Description
This project implements a Sudoku solver using Python. It utilizes a **Backtracking Algorithm**, which is a classic Computer Science technique for solving constraint-satisfaction problems. The algorithm attempts to place numbers 1 through 9 in empty cells, checking if each placement is valid according to Sudoku rules. If a placement leads to a dead end, it 'backtracks' (undoes the placement) and tries the next number.

```python
from typing import List, Tuple, Optional

def solve_sudoku(board: List[List[int]]) -> bool:
    empty = find_empty(board)
    if not empty:
        return True 
    
    row, col = empty
    
    for num in range(1, 10):
        if is_valid(board, num, (row, col)):
            board[row][col] = num
            if solve_sudoku(board):
                return True
            board[row][col] = 0
            
    return False

def is_valid(board: List[List[int]], num: int, pos: Tuple[int, int]) -> bool:
    row, col = pos
    for i in range(9):
        if board[row][i] == num and col != i: return False
    for i in range(9):
        if board[i][col] == num and row != i: return False
        
    box_x, box_y = col // 3, row // 3
    for i in range(box_y * 3, box_y * 3 + 3):
        for j in range(box_x * 3, box_x * 3 + 3):
            if board[i][j] == num and (i, j) != pos:
                return False
    return True

def find_empty(board: List[List[int]]) -> Optional[Tuple[int, int]]:
    for i in range(9):
        for j in range(9):
            if board[i][j] == 0: return (i, j)
    return None

if __name__ == "__main__":
    board = [
        [5, 3, 0, 0, 7, 0, 0, 0, 0],
        [6, 0, 0, 1, 9, 5, 0, 0, 0],
        [0, 9, 8, 0, 0, 0, 0, 6, 0],
        [8, 0, 0, 0, 6, 0, 0, 0, 3],
        [4, 0, 0, 8, 0, 3, 0, 0, 1],
        [7, 0, 0, 0, 2, 0, 0, 0, 6],
        [0, 6, 0, 0, 0, 0, 2, 8, 0],
        [0, 0, 0, 4, 1, 9, 0, 0, 5],
        [0, 0, 0, 0, 8, 0, 0, 7, 9]
    ]
    solve_sudoku(board)
    print(board)
```

### Project Explanation
The core of this solver is the recursive `solve_sudoku` function. It first scans the 9x9 grid to find an empty spot. It loops through numbers 1-9, trying to place each in the empty cell. Before placing a number, it calls `is_valid` to check if that number violates Sudoku rules. If a number is valid, it's placed on the board, and the function calls itself recursively. If a dead end is reached, it backtracks by resetting the cell to `0`.

### Project Remarks
* **Time Complexity:** The worst-case time complexity is O(9^m).
* **Real-World Application:** Backtracking is a fundamental algorithmic pattern used in pathfinding, scheduling, and resource allocation problems.
