# Project 6 Employee BST

### Project Description
Implements a Binary Search Tree (BST) Data Structure using Object-Oriented Programming to demonstrate O(log n) database indexing speeds.

```python
class EmployeeNode:
    def __init__(self, emp_id: int, name: str):
        self.emp_id = emp_id
        self.name = name
        self.left = None
        self.right = None

class EmployeeBST:
    def __init__(self):
        self.root = None

    def insert(self, emp_id: int, name: str):
        if not self.root:
            self.root = EmployeeNode(emp_id, name)
        else:
            self._insert(self.root, emp_id, name)

    def _insert(self, node, emp_id, name):
        if emp_id < node.emp_id:
            if not node.left: node.left = EmployeeNode(emp_id, name)
            else: self._insert(node.left, emp_id, name)
        elif emp_id > node.emp_id:
            if not node.right: node.right = EmployeeNode(emp_id, name)
            else: self._insert(node.right, emp_id, name)

    def search(self, emp_id: int):
        return self._search(self.root, emp_id)

    def _search(self, node, emp_id):
        if not node or node.emp_id == emp_id: return node
        if emp_id < node.emp_id: return self._search(node.left, emp_id)
        return self._search(node.right, emp_id)

    def inorder(self, node, result):
        if node:
            self.inorder(node.left, result)
            result.append(node.emp_id)
            self.inorder(node.right, result)

if __name__ == "__main__":
    db = EmployeeBST()
    for e_id, name in [(105, "Alice"), (102, "Bob"), (108, "Charlie")]:
        db.insert(e_id, name)
    
    res = db.search(108)
    print("Found:" if res else "Not Found", res.name if res else "")
```

### Project Explanation
Data is bundled into `Node` objects pointing to left/right children. Searching cuts the dataset in half at every step, bypassing the need to iterate line-by-line like a slow Python `for` loop.

### Project Remarks
This algorithmic logic is the foundation of production database indices (like PostgreSQL B-Trees).
